```python
from ultralytics import YOLO
from ultralytics.utils import SETTINGS

# SETTINGS.reset()
# print(SETTINGS)

model = YOLO("yolo26n.pt")  # load a pretrained model (recommended for training)

model.train(
    data="yolo_dataset.yaml",
    epochs=150,
    imgsz=64,
    workers=0,
)
```
